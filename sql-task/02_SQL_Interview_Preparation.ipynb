# 01. Introduction to SQL

## 1.1 What is SQL?

SQL (Structured Query Language) is a standard language used to store, retrieve, manipulate and manage data in relational database management systems (RDBMS).

### Applications of SQL
- Retrieving data
- Inserting records
- Updating records
- Deleting records
- Creating database structures
- Managing relationships between tables
- Controlling database transactions

### Popular RDBMS
- MySQL
- PostgreSQL
- Oracle Database
- Microsoft SQL Server
- SQLite


## 1.2 Database Fundamentals

### Database
An organized collection of related data.

### DBMS
Database Management System: Software used to create, store, retrieve and manage databases.

### RDBMS
Relational Database Management System: Stores data in related tables consisting of rows and columns.

### Basic Terminology

| Term | Meaning |
|---|---|
| Table | Collection of related records |
| Row | Individual record |
| Column | Attribute or field |
| Schema | Logical structure of a database |
| Primary Key | Uniquely identifies each row |
| Foreign Key | References a key in another table |
| NULL | Represents missing or unknown information |

### DBMS vs RDBMS

- DBMS is the broader category of database management software.
- RDBMS organizes data into related tables and typically enforces relationships using keys and constraints.


## 1.3 Primary Key and Foreign Key




### Primary Key
- Uniquely identifies each record.
- Cannot contain NULL.
- Must contain unique values.
- A table has only one primary key constraint, which may consist of multiple columns.

### Foreign Key
- Establishes a relationship between tables.
- References a primary key or another eligible unique key.
- Can contain NULL unless restricted by NOT NULL.
- Helps maintain referential integrity.

### Example

Employees Table:

| employee_id (PK) | name | department_id (FK) |
|---|---|---|
| 101 | Rahul | 1 |
| 102 | Amit | 2 |
| 103 | Priya | 1 |

Departments Table:

| department_id (PK) | department_name |
|---|---|
| 1 | IT |
| 2 | HR |

Here, `department_id` connects both tables.


```python
import sqlite3

# Create an in-memory database
conn = sqlite3.connect(":memory:")

# Enable foreign key enforcement
conn.execute("PRAGMA foreign_keys = ON")

cursor = conn.cursor()

print("SQLite database connected successfully!")
```

## Create sample tables

```python
cursor.execute("""
CREATE TABLE departments (
    department_id INTEGER PRIMARY KEY,
    department_name TEXT NOT NULL
)
""")

cursor.execute("""
CREATE TABLE employees (
    employee_id INTEGER PRIMARY KEY,
    employee_name TEXT NOT NULL,
    salary REAL,
    department_id INTEGER,
    FOREIGN KEY (department_id)
        REFERENCES departments(department_id)
)
""")

cursor.executemany("""
INSERT INTO departments VALUES (?, ?)
""", [
    (1, "IT"),
    (2, "HR"),
    (3, "Finance")
])

cursor.executemany("""
INSERT INTO employees VALUES (?, ?, ?, ?)
""", [
    (101, "Rahul", 50000, 1),
    (102, "Amit", 40000, 2),
    (103, "Priya", 60000, 1),
    (104, "Neha", 45000, 3)
])

conn.commit()

print("Sample tables created successfully!")
```

```python
#Retrieve sample data
cursor.execute("SELECT * FROM employees")

for row in cursor.fetchall():
    print(row)
```

# 2. SQL Commands

## 2.1 Categories of SQL Commands

SQL commands are commonly divided into five categories.

| Category | Full Form | Purpose |
|---|---|---|
| DDL | Data Definition Language | Defines database structures |
| DML | Data Manipulation Language | Inserts, updates and deletes data |
| DQL | Data Query Language | Retrieves data |
| DCL | Data Control Language | Controls database permissions |
| TCL | Transaction Control Language | Manages transactions |

### Important Commands

**DDL**
- CREATE
- ALTER
- DROP
- TRUNCATE

**DML**
- INSERT
- UPDATE
- DELETE

**DQL**
- SELECT

**DCL**
- GRANT
- REVOKE

**TCL**
- COMMIT
- ROLLBACK
- SAVEPOINT


## 2.2 DDL


### 2.2 DDL — Data Definition Language

DDL commands define or modify database structures.

### CREATE
Creates a new database object.

### ALTER
Modifies an existing database object.

### DROP
Removes a database object.

### TRUNCATE
Removes all rows from a table while retaining its structure in systems that support it.

**Interview trick:** SQLite does not support the TRUNCATE TABLE command. DELETE FROM table_name can remove all rows instead.


```python
# CREATE
cursor.execute("""
CREATE TABLE projects (
    project_id INTEGER PRIMARY KEY,
    project_name TEXT
)
""")

# ALTER
cursor.execute("""
ALTER TABLE projects
ADD COLUMN budget REAL
""")

# Verify table structure
cursor.execute("PRAGMA table_info(projects)")
print(cursor.fetchall())
```

## 2.3 DML — Data Manipulation Language


### 2.3 DML — Data Manipulation Language

DML commands manipulate records stored in tables.

### INSERT
Adds new records.

### UPDATE
Modifies existing records.

### DELETE
Removes records.

**Important:** Always use a WHERE condition when you intend to update or delete selected records.

Without WHERE, UPDATE and DELETE can affect every row.


```python
# INSERT
cursor.execute("""
INSERT INTO employees
VALUES (105, 'Arjun', 55000, 1)
""")

# UPDATE
cursor.execute("""
UPDATE employees
SET salary = 58000
WHERE employee_id = 105
""")

# DELETE
cursor.execute("""
DELETE FROM employees
WHERE employee_id = 105
""")

conn.commit()

print("DML operations completed.")
```

## 2.4 DQL — Data Query Language

DQL is used to retrieve data from database tables.

### SELECT

Syntax:

SELECT column1, column2
FROM table_name
WHERE condition;

### Examples

SELECT * FROM employees;

SELECT employee_name, salary
FROM employees;


```python
# Retrieve all employees
cursor.execute("SELECT * FROM employees")
print(cursor.fetchall())

# Retrieve selected columns
cursor.execute("""
SELECT employee_name, salary
FROM employees
""")

print(cursor.fetchall())
```

# DCL and TCL

## 2.5 DCL — Data Control Language

Controls database access and permissions.

- GRANT: Provides privileges.
- REVOKE: Removes privileges.

## 2.6 TCL — Transaction Control Language

Manages database transactions.

- COMMIT: Permanently saves transaction changes.
- ROLLBACK: Reverts uncommitted changes.
- SAVEPOINT: Creates a point within a transaction to which you can roll back.

**Important:** Exact command support and transaction behavior vary across database systems.


```python
# Start a transaction
cursor.execute("BEGIN")

cursor.execute("""
UPDATE employees
SET salary = 70000
WHERE employee_id = 101
""")

# Undo the update
conn.rollback()

# Verify original salary
cursor.execute("""
SELECT employee_name, salary
FROM employees
WHERE employee_id = 101
""")

print(cursor.fetchone())
```


## 2.7 Important SQL Interview Questions

**Q1. What is the difference between DDL and DML?**

DDL defines database structures; DML modifies table records.

**Q2. What is the difference between DELETE, TRUNCATE and DROP?**

DELETE removes selected or all rows. TRUNCATE removes all rows while retaining the table structure in supporting databases. DROP removes the database object itself.

**Q3. What is the difference between DELETE and DROP?**

DELETE removes records; DROP removes the database object.

**Q4. What is COMMIT?**

It makes transaction changes permanent.

**Q5. What is ROLLBACK?**

It undoes uncommitted transaction changes.

**Q6. What is SAVEPOINT?**

A marker within a transaction that allows partial rollback.

**Q7. Is SELECT DML or DQL?**

It is conventionally categorized as DQL, although some references group it under DML.

**Q8. Does SQLite support TRUNCATE?**

No. Use DELETE FROM table_name to remove all rows.

**Q9. What happens if UPDATE is executed without WHERE?**

Every row in the target table is updated.

**Q10. What happens if DELETE is executed without WHERE?**

Every row in the target table is deleted.
