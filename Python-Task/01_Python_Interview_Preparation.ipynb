## 1.2 How does Python execute a program?

Python execution involves several stages.

1. Source code is written in a `.py` file.
2. Python parses and compiles the source code into code objects.
3. CPython commonly generates bytecode.
4. The Python Virtual Machine (PVM) executes the bytecode.
5. Results are produced at runtime.

**Important:** Python is commonly called an interpreted language, but CPython also performs a compilation-to-bytecode step before execution.

## 2.4 Type Conversion in Python

Type conversion means converting a value from one data type to another.

Python supports two major types of conversion.

### 1. Implicit Type Conversion

Python automatically converts a value into another compatible type during certain operations.

Example:

```python
a = 10
b = 2.5

result = a + b

print(result)
print(type(result))
```

Output:

```text
12.5
<class 'float'>
```

Here, Python promotes the integer to a float for the arithmetic operation.

### 2. Explicit Type Conversion

Explicit conversion is performed by the programmer using built-in functions.

Common conversion functions:

| Function | Purpose |
|---|---|
| int() | Convert to integer |
| float() | Convert to float |
| str() | Convert to string |
| bool() | Convert to boolean |
| list() | Convert to list |
| tuple() | Convert to tuple |
| set() | Convert to set |
| dict() | Create a dictionary from suitable key-value pairs |

**Important:** Not every conversion is valid. For example, `int("hello")` raises a ValueError.


## Checking data types

### 2.5 type() and isinstance()

Python provides two useful functions for checking object types.

### type()

Returns the exact type of an object.

### isinstance()

Checks whether an object is an instance of a specified class or its subclass.

Example:

```python
number = 100

print(type(number))
print(isinstance(number, int))
```

Output:

```text
<class 'int'>
True
```

### Difference

- `type()` checks the exact type.
- `isinstance()` supports inheritance-aware type checking.

In production code, `isinstance()` is often more flexible when working with class hierarchies.


````python
# Integer to float
number = 10
converted_float = float(number)

# Float to integer
price = 99.99
converted_int = int(price)

# Integer to string
age = 23
converted_string = str(age)

# String to integer
value = "500"
converted_value = int(value)

print(converted_float, type(converted_float))
print(converted_int, type(converted_int))
print(converted_string, type(converted_string))
print(converted_value, type(converted_value))
````

````python
#type() vs isinstance()
class Animal:
    pass


class Dog(Animal):
    pass


dog = Dog()

print(type(dog) == Dog)
print(isinstance(dog, Dog))
print(isinstance(dog, Animal))
````

## Mutable and immutable objects


### 2.6 Mutable vs Immutable Data Types

This is one of the most important Python interview concepts.

### Mutable Objects

Mutable objects can be modified after creation without creating a new object.

Examples:

- list
- dict
- set
- bytearray

### Immutable Objects

Immutable objects cannot be modified after creation.

Examples:

- int
- float
- bool
- str
- tuple
- frozenset
- bytes

### Example

```python
numbers = [10, 20, 30]
numbers.append(40)

print(numbers)
```

The existing list is modified.

Now consider a string:

```python
name = "Python"
name = name + " Programming"

print(name)
```

A new string object is created for the concatenated result, and the variable is rebound to it.

**Interview point:** Mutability is a property of objects, not variable names.

### 2. Short-Circuit Evaluation

Python evaluates logical expressions from left to right.


## 4.2 Aliasing vs Copying



This is a frequently asked Python interview concept.

### Aliasing

Two variables refer to the same object.

### Shallow Copy

Creates a new outer collection, but nested objects are shared.

### Deep Copy

Creates an independent copy of nested objects as well.

Important functions:
- `copy()`
- `copy.copy()`
- `copy.deepcopy()`


## Shallow vs Deep Copy

````python
import copy

original = [[1, 2], [3, 4]]

shallow = copy.copy(original)
deep = copy.deepcopy(original)

original[0][0] = 100

print("Original:", original)
print("Shallow:", shallow)
print("Deep:", deep)
````

## 4.3 Comprehensions



Comprehensions provide a concise way to create collections.

Types:
- List comprehension
- Set comprehension
- Dictionary comprehension

General syntax:

`[expression for item in iterable if condition]`


````python
# List comprehension
squares = [x**2 for x in range(1, 6)]

# Set comprehension
unique_squares = {x**2 for x in [1, 2, 2, 3]}

# Dictionary comprehension
square_dict = {x: x**2 for x in range(1, 4)}

print(squares)
print(unique_squares)
print(square_dict)
````

## 5.1 What is a Function?

A function is a reusable block of code designed to perform a specific task.

## Types of arguments


#### 5.2 Types of Function Arguments

1. Positional arguments
2. Keyword arguments
3. Default arguments
4. Variable-length arguments

### Example

```python
def student(name, age=22):
    print(name, age)

student("Rahul")            # Default argument
student("Amit", 24)         # Positional arguments

```

**Important:** Required parameters without defaults must generally appear before parameters with defaults.


````python
#*args and **kwargs
# *args: Accepts multiple positional arguments

def calculate_total(*args):
    return sum(args)

print(calculate_total(10, 20, 30, 40))


# **kwargs: Accepts multiple keyword arguments

def display_details(**kwargs):
    for key, value in kwargs.items():
        print(key, ":", value)

display_details(name="Pintu", age=23, city="Kolkata")
````

Remember:
- *args collects extra positional arguments into a tuple.
- **kwargs collects extra keyword arguments into a dictionary.

## Return vs Print


### 5.3 Difference Between return and print()

**print():**
Displays information on the screen.

**return:**

If a function has no explicit return statement, it returns None.


````python
def function_a():
    print("Hello")

def function_b():
    return "Hello"

result_a = function_a()
result_b = function_b()

print("Result A:", result_a)
print("Result B:", result_b)
````

## 5.4 Scope of variables




Variable scope determines where a variable can be accessed.

### Types

- Local: Defined inside a function.
- Global: Defined at module level.
- Enclosing: Defined in an outer function.
- Built-in: Provided by Python.

Python follows the LEGB lookup rule:

1. Local
2. Enclosing
3. Global
4. Built-in

`global` allows assignment to a module-level variable.

`nonlocal` allows assignment to a variable in an enclosing function scope.


````python
x = 100

def outer():
    x = 50

    def inner():
        nonlocal x
        x = 25

    inner()
    print("Outer:", x)

outer()
print("Global:", x)
````

## Mutable default argument trap


## 5.5 Mutable Default Argument Trap

Python evaluates default argument expressions once when the function is defined, not every time it is called.

Using a mutable default such as a list can cause unexpected shared state between calls.

### Incorrect approach

```python
def add_item(item, items=[]):
    items.append(item)
    return items
```

The same list is reused across calls.

### Correct approach

```python
def add_item(item, items=None):
    if items is None:
        items = []

    items.append(item)
    return items
```

Use None when a fresh mutable default is required for each call.


````python
def add_item(item, items=[]):
    items.append(item)
    return items

print(add_item(10))
print(add_item(20))
print(add_item(30))
````
